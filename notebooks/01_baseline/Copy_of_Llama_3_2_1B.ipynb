```python
model_name = "meta-llama/Llama-3.2-1B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    device_map="auto"
)

print("Model loaded successfully")
```

```python
from datasets import load_dataset

dataset = load_dataset(
    "BioLaySumm/BioLaySumm2025-LaymanRRG-opensource-track"
)

print(dataset)
```

```python
print(dataset["train"].column_names)

print(dataset["train"][0])
```

```python
small_dataset = (
    dataset["train"]
    .shuffle(seed=42)
    .select(range(100))
)

print("Number of samples:", len(small_dataset))
```

```python
import random

sample = random.choice(small_dataset)

print("RADIOLOGY REPORT:\n")
print(sample["radiology_report"])

print("\nGROUND TRUTH LAYMAN SUMMARY:\n")
print(sample["layman_report"])
```

```python
report = sample["radiology_report"]

prompt = f"""
You are an expert medical communication assistant.

Task:
Convert the following radiology report into a patient-friendly explanation.

Instructions:
- Use simple language understandable by a non-medical person.
- Preserve all findings from the report.
- Do not add information that is not present in the report.
- Do not speculate or make new diagnoses.
- Explain medical terms in plain language when appropriate.
- Keep the summary concise (2–4 sentences).
- If the report is normal, clearly reassure the patient.
- Output only the summary.

Radiology Report:
{report}

Patient-Friendly Summary:
"""

inputs = tokenizer(
    prompt,
    return_tensors="pt",
    truncation=True,
    max_length=1024
).to(model.device)

outputs = model.generate(
    **inputs,
    max_new_tokens=120,
    do_sample=False,      # deterministic output
    temperature=0.0,
    pad_token_id=tokenizer.eos_token_id
)

# Extract only newly generated tokens
generated_tokens = outputs[0][inputs.input_ids.shape[1]:]

summary = tokenizer.decode(
    generated_tokens,
    skip_special_tokens=True
).strip()

print("PATIENT-FRIENDLY SUMMARY:\n")
print(summary)
```

```python
predictions = []
references = []
reports = []

for i, sample in enumerate(small_dataset.select(range(100))):
    print("Running sample:", i + 1)

    report = sample["radiology_report"]

    prompt = f"""
Rewrite this radiology report in simple patient-friendly language.
Use only the report information.
Output only 1 sentence.

Radiology Report:
{report}

Summary:
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=256
    ).to("cuda")

    with torch.inference_mode():
        outputs = model.generate(
            **inputs,
            max_new_tokens=30,
            do_sample=False,
            use_cache=True,
            pad_token_id=tokenizer.eos_token_id
        )

    summary = tokenizer.decode(
        outputs[0][inputs.input_ids.shape[1]:],
        skip_special_tokens=True
    ).strip()

    predictions.append(summary)
    references.append(sample["layman_report"])
    reports.append(report)

print("Generated:", len(predictions))
print(len(reports))
print(len(references))
print(len(predictions))
```

```python
import torch
import pandas as pd
from transformers import AutoTokenizer, AutoModelForSequenceClassification

nli_model_name = "MoritzLaurer/DeBERTa-v3-base-mnli-fever-anli"

nli_tokenizer = AutoTokenizer.from_pretrained(nli_model_name)
nli_model = AutoModelForSequenceClassification.from_pretrained(nli_model_name).to("cuda" if torch.cuda.is_available() else "cpu")

nli_model.eval()
```

```python
def check_hallucination(report, generated_summary):
    premise = str(report)
    hypothesis = str(generated_summary)

    inputs = nli_tokenizer(
        premise,
        hypothesis,
        return_tensors="pt",
        truncation=True,
        max_length=512
    ).to(nli_model.device)

    with torch.no_grad():
        outputs = nli_model(**inputs)

    probs = torch.softmax(outputs.logits, dim=-1)[0]

    labels = nli_model.config.id2label
    scores = {labels[i]: probs[i].item() for i in range(len(probs))}

    predicted_label = max(scores, key=scores.get)

    hallucinated = predicted_label != "entailment"

    return predicted_label, scores, hallucinated
```

```python
df = pd.read_csv("llama_results.csv")

print(df.head())
print(df.columns)
```

```python
nli_labels = []
hallucination_flags = []

for i, row in df.iterrows():
    label, scores, hallucinated = check_hallucination(
        row["Radiology Report"],
        row["Llama Generated Summary"]
    )

    nli_labels.append(label)
    hallucination_flags.append(hallucinated)

df["NLI_Label"] = nli_labels
df["Hallucinated"] = hallucination_flags

df.head()
```

```python
hallucination_count = df["Hallucinated"].sum()
total = len(df)

hallucination_rate = hallucination_count / total

print("Total samples:", total)
print("Hallucinated samples:", hallucination_count)
print("Hallucination rate:", hallucination_rate)
print("Hallucination percentage:", hallucination_rate * 100)
```

```python
shuffled_dataset = dataset["train"].shuffle(seed=42)

print(len(shuffled_dataset))
```

```python
comparison_dataset = shuffled_dataset.select(range(100,110))

print(len(comparison_dataset))
```

```python
comparison_reports = []
comparison_groundtruth = []
comparison_predictions = []
```

```python
for i, sample in enumerate(comparison_dataset):

    print(f"Generating Sample {i+1}")

    report = sample["radiology_report"]

    prompt = f"""
# Medical Text Simplification Assistant

## Persona
You are a skilled medical communication assistant.

## Task
Generate a clear and patient-friendly layman summary based on the provided radiology report.

## Steps
1. Review the radiology report carefully.
2. Rewrite the report in simple, understandable language.
3. Ensure the summary is:
   - Accurate
   - Patient-friendly
   - Medically relevant
   - Based only on the report
4. Exclude any additional commentary or repetition of the report.
5. Output only the layman summary.

## Radiology Report
{report}

## Layman Summary
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    ).to(model.device)

    with torch.inference_mode():
        outputs = model.generate(
            **inputs,
            max_new_tokens=80,
            do_sample=False,
            use_cache=True,
            pad_token_id=tokenizer.eos_token_id
        )

    summary = tokenizer.decode(
        outputs[0][inputs.input_ids.shape[1]:],
        skip_special_tokens=True
    ).strip()

    comparison_reports.append(report)
    comparison_groundtruth.append(sample["layman_report"])
    comparison_predictions.append(summary)

print("Generation completed!")
```

```python
comparison_df = pd.DataFrame({
    "Radiology Report": comparison_reports,
    "Ground Truth Summary": comparison_groundtruth,
    "Llama Generated Summary": comparison_predictions
})

comparison_df
```

```python
!pip install -q transformers sentencepiece accelerate

import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification

nli_model_name = "MoritzLaurer/DeBERTa-v3-base-mnli-fever-anli"

nli_tokenizer = AutoTokenizer.from_pretrained(nli_model_name)

nli_model = AutoModelForSequenceClassification.from_pretrained(
    nli_model_name
).to("cuda" if torch.cuda.is_available() else "cpu")

nli_model.eval()
```

```python
def check_hallucination(report, generated_summary):

    inputs = nli_tokenizer(
        str(report),
        str(generated_summary),
        return_tensors="pt",
        truncation=True,
        max_length=512
    ).to(nli_model.device)

    with torch.no_grad():
        outputs = nli_model(**inputs)

    probs = torch.softmax(outputs.logits, dim=-1)[0]

    labels = nli_model.config.id2label

    scores = {labels[i]: probs[i].item() for i in range(len(probs))}

    predicted_label = max(scores, key=scores.get)

    hallucinated = predicted_label != "entailment"

    return predicted_label, hallucinated
```

```python
hallucination_flags = []

for _, row in comparison_df.iterrows():

    _, hallucinated = check_hallucination(
        row["Radiology Report"],
        row["Llama Generated Summary"]
    )

    hallucination_flags.append(hallucinated)

comparison_df["Hallucinated"] = [
    "Yes" if x else "No"
    for x in hallucination_flags
]

comparison_df
```

```python
hallucination_count = sum(
    comparison_df["Hallucinated"] == "Yes"
)

total = len(comparison_df)

hallucination_rate = hallucination_count / total

print("Total Samples:", total)
print("Hallucinated Samples:", hallucination_count)
print(f"Hallucination Percentage: {hallucination_rate*100:.2f}%")
```
